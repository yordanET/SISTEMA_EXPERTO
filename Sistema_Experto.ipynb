# Sistema Experto para Diagnóstico de una Computadora




# 1. Knowledge Base — Base de Conocimientos



```python
# BASE DE CONOCIMIENTOS

base_conocimientos = [
    {
        "regla": "R1",
        "condiciones": {"no_enciende", "sin_luces"},
        "diagnostico": "Problema en la fuente de alimentación",
        "recomendacion": "Revisar cable de energía, tomacorriente y fuente de poder."
    },
    {
        "regla": "R2",
        "condiciones": {"enciende", "pantalla_negra"},
        "diagnostico": "Posible problema de memoria RAM o video",
        "recomendacion": "Revisar memoria RAM, cable del monitor y tarjeta gráfica."
    },
    {
        "regla": "R3",
        "condiciones": {"se_reinicia", "se_calienta"},
        "diagnostico": "Sobrecalentamiento del equipo",
        "recomendacion": "Limpiar ventiladores y revisar la pasta térmica."
    },
    {
        "regla": "R4",
        "condiciones": {"lento", "disco_100"},
        "diagnostico": "Problema de rendimiento del almacenamiento",
        "recomendacion": "Revisar procesos en segundo plano y el estado del HDD o SSD."
    },
    {
        "regla": "R5",
        "condiciones": {"sin_internet", "wifi_conectado"},
        "diagnostico": "Problema de red o configuración IP",
        "recomendacion": "Reiniciar el router y revisar la configuración de red."
    },
    {
        "regla": "R6",
        "condiciones": {"ruido_extraño", "lento"},
        "diagnostico": "Posible fallo del disco duro",
        "recomendacion": "Realizar una copia de seguridad y revisar el estado del disco."
    }
]

print("Base de conocimientos cargada correctamente.")
print("Número de reglas:", len(base_conocimientos))
```

# 2. Inference Engine — Motor de Inferencia


```python
# MOTOR DE INFERENCIA

def motor_inferencia(sintomas_usuario):
    sintomas_usuario = set(sintomas_usuario)
    resultados = []

    for regla in base_conocimientos:
        if regla["condiciones"].issubset(sintomas_usuario):
            resultados.append(regla)

    return resultados


def generar_respuesta(sintomas_usuario):
    resultados = motor_inferencia(sintomas_usuario)

    if not resultados:
        return (
            "No se encontró una regla que coincida completamente con los síntomas.\n"
            "Se recomienda realizar una revisión técnica más detallada."
        )

    texto = "RESULTADO DEL SISTEMA EXPERTO\n"
    texto += "=" * 40 + "\n\n"

    for i, regla in enumerate(resultados, start=1):
        texto += f"Diagnóstico {i}: {regla['diagnostico']}\n"
        texto += f"Regla aplicada: {regla['regla']}\n"
        texto += "Síntomas relacionados: "
        texto += ", ".join(sorted(regla["condiciones"])) + "\n"
        texto += f"Recomendación: {regla['recomendacion']}\n"
        texto += "-" * 40 + "\n"

    return texto
```

## Prueba rápida del motor de inferencia



```python
sintomas_prueba = ["se_reinicia", "se_calienta"]

print("Síntomas ingresados:", sintomas_prueba)
print()
print(generar_respuesta(sintomas_prueba))
```

# 3. User Interface — Interfaz de Usuario


```python
# Si ipywidgets no está instalado, descomenta la siguiente línea:
# !pip install ipywidgets

import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
# OPCIONES DE LA INTERFAZ

sintomas_disponibles = {
    "La computadora no enciende": "no_enciende",
    "No aparecen luces o indicadores": "sin_luces",
    "La computadora enciende": "enciende",
    "La pantalla permanece negra": "pantalla_negra",
    "La computadora se reinicia": "se_reinicia",
    "La computadora se calienta demasiado": "se_calienta",
    "La computadora está lenta": "lento",
    "El uso del disco llega al 100%": "disco_100",
    "No hay acceso a Internet": "sin_internet",
    "El WiFi aparece conectado": "wifi_conectado",
    "El disco hace ruidos extraños": "ruido_extraño"
}

selector_sintomas = widgets.SelectMultiple(
    options=[(nombre, codigo) for nombre, codigo in sintomas_disponibles.items()],
    description="Síntomas:",
    rows=11,
    layout=widgets.Layout(width="700px")
)

boton_diagnosticar = widgets.Button(
    description="Diagnosticar",
    button_style="success",
    icon="search"
)

boton_limpiar = widgets.Button(
    description="Limpiar",
    button_style="warning",
    icon="trash"
)

salida = widgets.Output()
```

```python
# FUNCIONES DE LA INTERFAZ

def diagnosticar(b):
    with salida:
        clear_output()

        sintomas_seleccionados = list(selector_sintomas.value)

        if len(sintomas_seleccionados) == 0:
            print("Seleccione al menos un síntoma.")
            return

        respuesta = generar_respuesta(sintomas_seleccionados)
        print(respuesta)


def limpiar(b):
    selector_sintomas.value = ()

    with salida:
        clear_output()
        print("Selección limpiada.")


boton_diagnosticar.on_click(diagnosticar)
boton_limpiar.on_click(limpiar)
```

```python
# MOSTRAR LA INTERFAZ

titulo = widgets.HTML(
    value="<h2>🖥️ Sistema Experto de Diagnóstico de Computadora</h2>"
)

descripcion = widgets.HTML(
    value=(
        "<p>Seleccione los síntomas que presenta la computadora "
        "y presione <b>Diagnosticar</b>.</p>"
    )
)

botones = widgets.HBox([
    boton_diagnosticar,
    boton_limpiar
])

interfaz = widgets.VBox([
    titulo,
    descripcion,
    selector_sintomas,
    botones,
    salida
])

display(interfaz)
```
